```python
import time

def linear_search(arr, key):
    for i in range(len(arr)):
        if arr[i] == key:
            return i
    return -1

def binary_search(arr, key):
    low, high = 0, len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == key:
            return mid
        elif arr[mid] < key:
            low = mid + 1
        else:
            high = mid - 1
    return -1

n = 100000
arr = [i + 1 for i in range(n)] # sorted array: [1, 2, 3,..., 100000]

key = int(input("Enter element to search: "))

# Linear Search
start = time.time()
index = linear_search(arr, key)
end = time.time()
print("\nLinear Search")
if index!= -1:
    print("Element found at index", index)
else:
    print("Element not found")
print("Time Taken :", round((end - start) * 1000000), "microseconds")

# Binary Search
start = time.time()
index = binary_search(arr, key)
end = time.time()
print("\nBinary Search")
if index!= -1:
    print("Element found at index", index)
else:
    print("Element not found")
print("Time Taken :", round((end - start) * 1000000), "microseconds")
```
