```python
import time

# Recursive function for factorial
def factorial_recursive(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial_recursive(n - 1)

# Iterative function for factorial
def factorial_iterative(n):
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

def main():
    try:
        n = int(input("Enter a non-negative integer (e.g., 20): "))
    except ValueError:
        print("Please enter a valid integer.")
        return

    if n < 0:
        print("Please enter a non-negative integer.")
        return

    # --- Iterative ---
    start1 = time.perf_counter_ns()
    iter_result = factorial_iterative(n)
    end1 = time.perf_counter_ns()
    iter_time = end1 - start1

    # --- Recursive ---
    start2 = time.perf_counter_ns()
    rec_result = factorial_recursive(n)
    end2 = time.perf_counter_ns()
    rec_time = end2 - start2

    # Output
    print(f"\n--- Results for {n}! ---")
    print(f"Iterative Result : {iter_result}")
    print(f"Iterative Time   : {iter_time} ns")
    print(f"------------------------------------")
    print(f"Recursive Result : {rec_result}")
    print(f"Recursive Time   : {rec_time} ns")

    print("\n=== Code Execution Successful ===")

if __name__ == "__main__":
    main()
```
